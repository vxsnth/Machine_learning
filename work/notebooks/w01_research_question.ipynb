### What the starter data suggests

The starter dataset contains 30,000 unique pages. About 54.2% of the pages are trending down, and the median page has 731 impressions over the past 90 days. These numbers suggest there is a meaningful set of pages that could be reviewed and prioritized. A ranking could help a content team focus limited review time on pages that may deserve attention. These numbers do not prove that refreshing a page will improve its performance, but they make the content opportunity question worth investigating further.

```python
import pandas as pd

url = "https://raw.githubusercontent.com/vxsnth/Machine_learning/main/data/raw/content_refresh_anonymized.csv"

df = pd.read_csv(url)

print("Dataset loaded successfully!")
print("Rows:", len(df))
print("Columns:", len(df.columns))

print("\n--- 3 real numbers ---")

print("Unique pages:", df["content_id"].nunique())

print(
    "Percentage trending down:",
    round((df["trend_direction"] == "down").mean() * 100, 1),
    "%"
)

print(
    "Median 90-day impressions:",
    round(df["impressions_90d"].median(), 0)
)
```
