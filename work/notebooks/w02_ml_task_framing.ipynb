### Why ML beats a fixed rule here

A fixed rule could rank pages using one threshold, such as reviewing every page whose impressions are below a chosen value. That would be simple and easy to explain, but it would ignore the fact that several page-level signals may matter at the same time.

For example, a page with lower impressions may still be worth protecting, while another page with higher impressions may show signals that make it a stronger review candidate. A ranking model can combine multiple observed signals and learn patterns associated with the target or proxy.

I would still compare the ML ranking with a simple rule-based baseline. ML is useful only if it provides better decision support than that simpler approach.


```python

threshold = df["impressions_90d"].median()

rule_candidates = df[
    df["impressions_90d"] < threshold
]

print("Simple rule: review pages below median 90-day impressions")
print("Threshold:", threshold)
print("Pages selected:", len(rule_candidates))
```
